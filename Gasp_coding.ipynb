```python

Largeur = 600
Hauteur = 500


class Brique: #définition de la brique, le rectangle qui sera détruit par la balle

    def __init__(self, canvas, x, y, largeur, hauteur, couleur):
        self.x = x
        self.y = y
        self.l = largeur
        self.h = hauteur
        self.canvas = canvas
        self.graph = canvas.create_rectangle(x, y, x + largeur, y + hauteur, fill=couleur)

    def supprimer(self): #sera utilisé pour supprimer la brique lorsqu'elle est touchée par la balle
        self.canvas.delete(self.graph)

class Raquette: 

    def __init__(self, canvas,):
        self.canvas = canvas
        self.l = 90
        self.h = 12
        self.vitesse = 8
        self.x = (Largeur - self.l) / 2
        self.y = Hauteur - 40
        self.graph = canvas.create_rectangle(0, 0, 0, 0, fill = "#e74c3c")
        self.placement()

    def placement(self): #on place la raquette qui était en position nulle à la bonne position
        self.canvas.coords(self.graph, self.x, self.y, self.x + self.l, self.y + self.h)
    
    def deplacer(self, input):
        if "Gauche" in input:
            self.x -= self.vitesse
        if "Droite" in input:
            self.x += self.vitesse
        self.x = max(0, min(Largeur-self.l, self.x)) #on bloque la raquette dans l'espace de jeu
        self.placement() #on replace la raquette à sa nouvelle position 



```
